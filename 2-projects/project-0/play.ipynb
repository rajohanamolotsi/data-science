### Warm-up

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df_0 = pd.read_csv('NJ.csv')
```

```python
df_0
```

```python
plt.scatter(df_0['area'], df_0['price'], marker='*', color='r')
plt.xlabel('area')
plt.ylabel('price')
plt.title('price vs. area in NJ')
plt.tight_layout()
```

```python
model = linear_model.LinearRegression()
```

```python
model.fit(df_0[['area']], df_0['price'])
```

```python
model.predict([[3300]])
```

```python
float(model.predict([[3300]]))
```
